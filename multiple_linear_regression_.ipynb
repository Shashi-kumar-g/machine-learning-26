```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# Step 1: Create Dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9],
    'Attendance': [60, 65, 70, 75, 80, 85, 90, 95],
    'Assignment_Marks': [50, 55, 60, 65, 70, 75, 85, 90],
    'Exam_Marks': [45, 50, 60, 65, 75, 80, 90, 95]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Step 2: Independent and Dependent Variables
X = df[['Study_Hours', 'Attendance', 'Assignment_Marks']]
Y = df['Exam_Marks']

# Step 3: Split Dataset
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.25, random_state=42
)

# Step 4: Create Model
model = LinearRegression()

# Step 5: Train Model
model.fit(X_train, Y_train)

# Step 6: Predict Values
Y_pred = model.predict(X_test)

print("\nActual Values:", list(Y_test))
print("Predicted Values:", Y_pred)

# Step 7: Model Coefficients
print("\nIntercept:", model.intercept_)
print("Coefficients:", model.coef_)

# Step 8: Evaluate Model
mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)

print("\nMean Squared Error:", mse)
print("R2 Score:", r2)

# Step 9: Predict New Student Marks
new_student = pd.DataFrame({
    'Study_Hours': [6],
    'Attendance': [85],
    'Assignment_Marks': [80]
})

prediction = model.predict(new_student)

print("\nPredicted Exam Marks:", prediction[0])

# Step 10: Actual vs Predicted Graph
plt.scatter(Y_test, Y_pred)
plt.xlabel("Actual Marks")
plt.ylabel("Predicted Marks")
plt.title("Multiple Linear Regression")
plt.show()
```
