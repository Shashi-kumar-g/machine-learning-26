```python
X = df.drop(columns=["defaulted_12m"])
y = df["defaulted_12m"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X = X.drop(columns=["loan_id", "applicant_name"], errors="ignore")

print(X.columns.tolist())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

numeric_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object"]).columns.tolist()

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", GradientBoostingClassifier(random_state=42))
])

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    model,
    X,
    y,
    cv=cv,
    scoring="roc_auc"
)

print("AUC scores for 5 folds:")
print(scores)

print("\nMean 5-fold AUC:")
print(scores.mean())
```

```python
leaky_features = [
    "collection_calls",
    "recovery_amount",
    "account_closed_date",
    "late_but_repaid"
]

X_no_leak = df.drop(
    columns=["defaulted_12m", "loan_id", "applicant_name"] + leaky_features
)

print("Remaining features:")
print(X_no_leak.columns.tolist())
```

```python
numeric_features_no_leak = X_no_leak.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features_no_leak = X_no_leak.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numeric:")
print(numeric_features_no_leak)

print("\nCategorical:")
print(categorical_features_no_leak)
```

```python
numeric_transformer_no_leak = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer_no_leak = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_no_leak = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_no_leak, numeric_features_no_leak),
        ("cat", categorical_transformer_no_leak, categorical_features_no_leak)
    ]
)
```

```python
model_no_leak = Pipeline(steps=[
    ("preprocessor", preprocessor_no_leak),
    ("classifier", GradientBoostingClassifier(random_state=42))
])

scores_no_leak = cross_val_score(
    model_no_leak,
    X_no_leak,
    y,
    cv=cv,
    scoring="roc_auc"
)

print("AUC scores after removing leakage:")
print(scores_no_leak)

print("\nMean 5-fold AUC after removing leakage:")
print(scores_no_leak.mean())
```

```python
leakage_features = [
    "collection_calls",
    "recovery_amount",
    "account_closed_date",
    "late_but_repaid",
    "risk_flag"
]

leakage_results = []

for feature in leakage_features:

    X_test = df.drop(
        columns=["defaulted_12m", "loan_id", "applicant_name", feature]
    )

    numeric_features_test = X_test.select_dtypes(
        include=["int64", "float64"]
    ).columns.tolist()

    categorical_features_test = X_test.select_dtypes(
        include=["object"]
    ).columns.tolist()

    preprocessor_test = ColumnTransformer(
        transformers=[
            (
                "num",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="median"))
                ]),
                numeric_features_test
            ),
            (
                "cat",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="most_frequent")),
                    ("onehot", OneHotEncoder(handle_unknown="ignore"))
                ]),
                categorical_features_test
            )
        ]
    )

    model_test = Pipeline([
        ("preprocessor", preprocessor_test),
        ("classifier", GradientBoostingClassifier(random_state=42))
    ])

    scores_test = cross_val_score(
        model_test,
        X_test,
        y,
        cv=cv,
        scoring="roc_auc"
    )

    leakage_results.append({
        "Feature Removed": feature,
        "Mean 5-Fold AUC": scores_test.mean()
    })

leakage_table = pd.DataFrame(leakage_results)

leakage_table
```
